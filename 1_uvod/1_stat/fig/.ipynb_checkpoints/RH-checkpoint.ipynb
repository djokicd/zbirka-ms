```python
RH = np.array([20, 40, 60, 80])
R = np.array([5e6, 310e3, 31e3, 5.7e3])
```

```python
w = 1.6
a = 1.618
figure (figsize = (a*w, w), dpi = 200)
grid(b=True, which='major', color='0.45', linestyle='-')
grid(b=True, which='minor', color='0.75', linestyle='--')
gca().xaxis.set_major_formatter(mticker.ScalarFormatter())

#gca().yaxis.set_major_formatter(mpl.ticker.FormatStrFormatter('%.2f%%'))
#minorticks_off()
plt.loglog(RH, R, 'x')
ylabel(r"$R\,[\Omega]$")
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

# Generate some sample data that spans several orders of magnitude
x = RH # Values from 10^1 to 10^5
y = R

# Create the plot
fig, ax = plt.subplots()
ax.loglog(x, y) # Create the log-log plot

# Change the formatter to display plain numbers instead of scientific notation
ax.xaxis.set_major_formatter(mticker.ScalarFormatter())
# Optionally, you can also set the minor formatter to NullFormatter
# to avoid minor ticks from showing exponents if needed
ax.xaxis.set_minor_formatter(mticker.NullFormatter())

# Ensure the formatter treats values as "plain" style (avoids scientific notation even for large numbers)
ax.ticklabel_format(style='plain', axis='x')

# Optional: Add grid and title for clarity
ax.grid(True, which="both", ls="-", color='0.90')
ax.set(title='Log-Log Plot with Plain X-axis Labels', xlabel='X Value', ylabel='Y Value (log scale)')

plt.show()
```
