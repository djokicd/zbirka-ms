```python
from ms_lib import *

R0 = 10e3 
beta = 3950
R = 1.5e3
T0 = 298

def f(T):
    return R - R0*exp( beta*(1/T - 1/T0) )

def fp(T):
    return beta*R0/(T**2)*exp( beta*(1/T - 1/T0) ) 
```

```python
f(298), fp(298)
```

```python
T = 298

T_iter = []

for k in range(4):
    T_iter += [T]
    T_next = T - f(T)/fp(T)
    T = T_next
    print(T_next)
```

```python
f(301.699)
```

```python
generate_canvas()
T_vec = np.linspace(T0, T0+100)
f_vec = [f(T) for T in T_vec]
f_iter = [f(T) for T in T_iter]

plt.plot(T_vec, f_vec)
plt.plot(T_iter, f_iter, 'r+', label="Итерације")
plt.xlabel(r"$T\rm\,[K]$")
plt.xlabel(r"$f(T)\rm\,[k\Omega/K]$")
plt.legend()
plt.savefig("nr_iter.pdf", bbox_inches = 'tight')
```
