```python
from pylab import *
from ms_lib import *
```

```python
n = 1e4
C = 14.07e-9 
L = 180e-6
R = 1e3

w = np.logspace(1, 10, 1000)
s = 1j*w
H = 1/n * s/C / (s**2 + s/(R*C) + 1/(L*C))
```

```python
W = 1.6
a = 1.618
figure (figsize = (a*W, W), dpi = 200)

grid(b=True, which='major', color='0.75', linestyle='--')
grid(b=True, which='minor', color='0.75', linestyle='--')

gca().xaxis.set_minor_locator(mticker.LogLocator(subs=np.arange(1, 10)))
gca().yaxis.set_minor_locator(mticker.LogLocator(subs=np.arange(1, 10)))
gca().minorticks_on() #

plt.loglog(w/(2*pi), abs(H), 'k' )
ylabel(r"$|R_\mathrm{m}({\rm j}\upomega)|\rm\,[\Omega]$")
xlabel(r"$f\rm\,[Hz]$")
plt.savefig("../../2_din/fig/H_jw.pdf", bbox_inches='tight')
```
