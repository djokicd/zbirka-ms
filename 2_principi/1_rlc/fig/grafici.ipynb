```python
generate_canvas()
ylim([0, 1.3])

C0 = 0.016
C1 = 1.24
C2 = 0.96

x = np.array([0, 1])
y1 = np.array([C1, C0])
y2 = np.array([C2, C0])
plt.plot(x, y1, 'k', label="$t_1$")
plt.plot(x, y2, 'k--', label="$t_2$")
plt.legend()
plt.xlabel(r"$x/\ell$")
plt.ylabel(r"C\rm\,[nF]")
plt.savefig("varC_sensor.pdf", bbox_inches='tight')
```

```python
generate_canvas()
mu0 = 4*pi*1e-7
mur = 100
N = 100
l = 50e-2
S = (2e-2)**2 
d_vec = np.linspace(0.1e-3, 3e-3, 1000)
B = mu0*mur*N**2*S/(l + 2*mur*d_vec)
plt.plot(d_vec/1e-3, B/1e-6, 'k')
xlabel(r"$d\rm\,[mm]$")
ylabel(r"$L\rm\,[\upmu H]$")
plt.savefig("L_rez.pdf", bbox_inches='tight')
```
